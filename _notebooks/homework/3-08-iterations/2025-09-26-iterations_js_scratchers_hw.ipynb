### 🧩 Task #1
Write a `for` loop that prints the numbers **1 through 10**.

````python
%%js
// Task 1: Print numbers 1 through 10
for (let i = 1; i <= 10; i++) {
    console.log(i);
}
````

### 🧩 Task #2
Write a `while` loop that prints the even numbers between **2 and 20**.

````python
%%js
// Task 2: Print even numbers between 2 and 20 using while loop
let num = 2;
while (num <= 20) {
    console.log(num);
    num += 2;
}
````

### 🧩 Task #3
Make the following shape using iteration loops with the symbol `*`:
```
*
**
***
****
*****
```

````python
%%js
// Task 3: Print a triangle of stars
let stars = "";
for (let i = 1; i <= 5; i++) {
    stars += "*";
    console.log(stars);
}
````

### 🧩 Task #4
Use a loop to calculate the **sum of all numbers from 1 to 100**, and print the result.

````python
%%js
// Task 4: Sum of numbers from 1 to 100
let sum = 0;
for (let i = 1; i <= 100; i++) {
    sum += i;
}
console.log("The sum is:", sum);
````
