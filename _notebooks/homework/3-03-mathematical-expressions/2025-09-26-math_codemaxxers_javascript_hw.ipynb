## Q1 (Exponents):
### A cube has a side length of 6 units. What is its volume?

```python
// Q1: Exponents — volume of a cube with side length 6
const side = 6;
const volume = side ** 3; // same as Math.pow(side, 3)
console.log('Q1 - Volume of cube (side 6):', volume);
volume;
```

## Q2 (PEMDAS):
