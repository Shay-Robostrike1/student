### Kali Linux laptops
The surplus laptops we got were slow and inefficient with windows 11 running, So I helped image them with linux
The process was relatively simple

Step 1. Set up imaging tools. Download the .iso file and an etcher for the usb drive. Then etch the drive with your chosen OS
Step 2. Preparing the computer. Turn on the computer and hold the key to open BIOS. It can be F10, F2, or del. 
Step 3. Boot from the USB drive. You can do this under an option in the BIOS labeled something along the lines of "boot from" or "secure bootable drive"
Step 4. Install tools and go!
